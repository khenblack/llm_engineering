```python
# imports
import os
from dotenv import load_dotenv
from openai import OpenAI
from IPython.display import Markdown, display, update_display
```

```python
# constants

MODEL_GPT = 'gpt-4o-mini'
MODEL_LLAMA = 'llama3.2'
```

```python
# set up environment
load_dotenv(override=True)
```

```python
# Setup Ollama
ollama_base_url = os.getenv("OLLAMA_BASE_URL")
model = "llama3.2"

ollama_client = OpenAI(base_url=ollama_base_url)
```

```python
# setup openai client
api_key = os.getenv("OPENAI_API_KEY")

if api_key and api_key.startswith('sk-proj-') and len(api_key)>10:
    print("API key looks good so far")
else:
    print("There might be a problem with your API key? Please visit the troubleshooting notebook!")

openai_client = OpenAI(api_key=api_key)    
```

```python
# System prompt
system_prompt = """
You are a teachnical tutor. Your responsibility is to help the user understand the technical code provided assuming they have no prior knowledge of the subject. Respond using Markdown syntax
"""

# User prompt
user_prompt = """
Here is the code provided by the user, please explain what the code does and why.
"""
```

```python
def explain_code(code):
    stream = ollama_client.chat.completions.create(
        model=MODEL_LLAMA,
        messages=[
            {"role": "system", "content": system_prompt},
            {"role": "user", "content": user_prompt + code}
        ],
        stream=True
    )

    response = ""
    display_handle = display(Markdown(""), display_id=True) 
    for chunk in stream:
        response += chunk.choices[0].delta.content or ''
        update_display(Markdown(response), display_id=display_handle.display_id)
    
```

```python
def pretty_explanation(code):
    explanation = explain_code(code)
    display(Markdown(explanation))
```
